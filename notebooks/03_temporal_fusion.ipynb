```python
import os
import xarray as xr
import numpy as np

def get_center(path):
    with xr.open_dataset(path) as ds:
        return (
            float(ds.attrs.get("center_lon")),
            float(ds.attrs.get("center_lat"))
        )

# Collect all Chimanimani files
s1_files = [
    f for f in os.listdir(extract_path_s1)
    if f.startswith("chimanimani_s1asc_")
]

s2_files = [
    f for f in os.listdir(nc_dir)
    if f.startswith("chimanimani_s2_")
]

# Read centers
s1_centers = {}
for f in s1_files:
    try:
        s1_centers[f] = get_center(os.path.join(extract_path_s1, f))
    except:
        pass

s2_centers = {}
for f in s2_files:
    try:
        s2_centers[f] = get_center(os.path.join(nc_dir, f))
    except:
        pass

print("S1 tiles with coordinates:", len(s1_centers))
print("S2 tiles with coordinates:", len(s2_centers))

# Find exact coordinate matches
matches = []

for s1_file, s1_center in s1_centers.items():
    for s2_file, s2_center in s2_centers.items():

        if (
            np.isclose(s1_center[0], s2_center[0]) and
            np.isclose(s1_center[1], s2_center[1])
        ):
            matches.append(
                (s1_file, s2_file, s1_center)
            )

print("\nExact spatial matches:", len(matches))

print("\nFirst 20 matches:")
for m in matches[:20]:
    print(m)
```

```python
import os
import numpy as np
import xarray as xr
import gc

def build_temporal_fused_dataset(
    s1_folder,
    s2_folder,
    region_prefix='chimanimani',
    limit=500
):
    X = []
    y = []

    # --------------------------------------------------
    # 1. Get S1 and S2 files
    # --------------------------------------------------
    s1_files = {
        f.replace('_s1asc_', '_s2_'): f
        for f in os.listdir(s1_folder)
        if f.startswith(region_prefix + '_s1asc_')
    }

    s2_files = {
        f: f
        for f in os.listdir(s2_folder)
        if f.startswith(region_prefix + '_s2_')
    }

    # --------------------------------------------------
    # 2. Match using the common filename ID
    # --------------------------------------------------
    common_files = sorted(set(s1_files.keys()) & set(s2_files.keys()))

    print("Matched S1-S2 files:", len(common_files))
    print("Using:", min(len(common_files), limit))

    # --------------------------------------------------
    # 3. Process matched pairs
    # --------------------------------------------------
    for idx, s2_name in enumerate(common_files[:limit]):

        s1_name = s1_files[s2_name]

        s1_path = os.path.join(s1_folder, s1_name)
        s2_path = os.path.join(s2_folder, s2_name)

        try:

            ds1 = xr.open_dataset(s1_path)
            ds2 = xr.open_dataset(s2_path)

            # ------------------------------------------
            # Verify temporal dimension
            # ------------------------------------------
            if ds1.sizes['time'] != 15 or ds2.sizes['time'] != 15:
                print("Skipping:", s1_name, "because time != 15")
                ds1.close()
                ds2.close()
                continue

            # ------------------------------------------
            # Extract temporal S1
            # ------------------------------------------
            vv = ds1['VV'].values.astype(np.float32)
            vh = ds1['VH'].values.astype(np.float32)

            # ------------------------------------------
            # Extract temporal S2
            # ------------------------------------------
            r = ds2['B04'].values.astype(np.float32)
            g = ds2['B03'].values.astype(np.float32)
            b = ds2['B02'].values.astype(np.float32)

            # ------------------------------------------
            # Normalize optical
            # Same basic scaling used in fusion dataset
            # ------------------------------------------
            optical = np.clip(
                np.stack([r, g, b], axis=-1) / 3000.0,
                0,
                1
            )

            # ------------------------------------------
            # Normalize radar
            # ------------------------------------------
            vv_norm = np.empty_like(vv)
            vh_norm = np.empty_like(vh)

            for t in range(15):

                vv_t = vv[t]
                vh_t = vh[t]

                # adaptive normalization
                vv_min = np.nanpercentile(vv_t, 2)
                vv_max = np.nanpercentile(vv_t, 98)

                vh_min = np.nanpercentile(vh_t, 2)
                vh_max = np.nanpercentile(vh_t, 98)

                vv_norm[t] = np.clip(
                    (vv_t - vv_min) / (vv_max - vv_min + 1e-8),
                    0,
                    1
                )

                vh_norm[t] = np.clip(
                    (vh_t - vh_min) / (vh_max - vh_min + 1e-8),
                    0,
                    1
                )

            # ------------------------------------------
            # Merge:
            # R G B VV VH
            #
            # Shape:
            # (15, 128, 128, 5)
            # ------------------------------------------
            fused = np.concatenate(
                [
                    optical,
                    vv_norm[..., np.newaxis],
                    vh_norm[..., np.newaxis]
                ],
                axis=-1
            )

            # ------------------------------------------
            # Ground-truth mask
            #
            # Use S2 MASK from the post-event observation
            # ------------------------------------------
            pre_post = ds2.attrs.get('pre_post_dates', {})

            if isinstance(pre_post, dict) and 'post' in pre_post:
                post_idx = int(pre_post['post'])
            else:
                post_idx = 14

            mask = ds2['MASK'].isel(
                time=post_idx
            ).values.astype(np.float32)

            # ------------------------------------------
            # Store
            # ------------------------------------------
            X.append(fused)
            y.append(mask[..., np.newaxis])

            ds1.close()
            ds2.close()

            if (idx + 1) % 50 == 0:
                print(
                    f"Processed {idx + 1}/{min(len(common_files), limit)}"
                )

        except Exception as e:

            print("Error:", s1_name, s2_name)
            print("Reason:", e)

            try:
                ds1.close()
                ds2.close()
            except:
                pass

            continue

    # --------------------------------------------------
    # Convert to numpy
    # --------------------------------------------------
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.float32)

    gc.collect()

    return X, y


# ======================================================
# BUILD TEMPORAL FUSED DATASET
# ======================================================

X_temporal_fused, y_temporal_fused = build_temporal_fused_dataset(
    extract_path_s1,
    nc_dir,
    region_prefix='chimanimani',
    limit=500
)

print("\n===================================")
print("TEMPORAL FUSED DATASET")
print("===================================")

print("X shape:", X_temporal_fused.shape)
print("y shape:", y_temporal_fused.shape)

print("X dtype:", X_temporal_fused.dtype)
print("y dtype:", y_temporal_fused.dtype)
```

```python
import numpy as np

X_path = "/content/data/temporal_fused/X_temporal_fused.npy"
y_path = "/content/data/temporal_fused/y_temporal_fused.npy"

# Memory-map instead of loading another 2.3 GB copy into RAM
X_check = np.load(X_path, mmap_mode='r')
y_check = np.load(y_path, mmap_mode='r')

print("========== SHAPES ==========")
print("X:", X_check.shape)
print("y:", y_check.shape)

print("\n========== VALUE RANGES ==========")

print("X min:", np.nanmin(X_check))
print("X max:", np.nanmax(X_check))
print("X mean:", np.nanmean(X_check))

print("\n========== CHANNEL CHECK ==========")

channel_names = ["R", "G", "B", "VV", "VH"]

for i, name in enumerate(channel_names):
    channel = X_check[:, :, :, :, i]

    print(
        f"{name}: "
        f"min={np.nanmin(channel):.4f}, "
        f"max={np.nanmax(channel):.4f}, "
        f"mean={np.nanmean(channel):.4f}"
    )

print("\n========== MASK CHECK ==========")

print("Mask min:", np.nanmin(y_check))
print("Mask max:", np.nanmax(y_check))
print("Mask mean:", np.nanmean(y_check))

# Percentage of samples containing at least one landslide pixel
sample_has_landslide = np.any(y_check > 0, axis=(1, 2, 3))

print(
    "Samples containing landslide pixels:",
    np.sum(sample_has_landslide),
    "/",
    len(sample_has_landslide)
)

print(
    "Percentage:",
    100 * np.mean(sample_has_landslide),
    "%"
)
```
